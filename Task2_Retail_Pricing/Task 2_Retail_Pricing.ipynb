```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    'Product' :['Milk','Bread','Yogurt','Bottled Water','Soda'],
    'Cost Price' :[40,25,30,10,35],
    'Recommended Price' :[56,36,46,16,51]
}

df = pd.DataFrame(data)

plt.figure(figsize=(8,5))
plt.bar(df['Product'], df['Recommended Price'])

plt.title('Recommended Retail Prices')
plt.xlabel('Product')
plt.ylabel('Price')

plt.show()
```

```python
plt.savefig('Retail_Prices.png', bbox_inches='tight')
plt.show()
```

## Key Findings :
1. Milk has the highest recommended retail price among the selected products.
2. Bottled Water has the lowest recommended price due to strong market competition.
3. All recommended prices are higher than cost prices, ensuring profitability.
4. Competitor pricing was considered while setting the final recommended price.
5. Proper pricing helps businesses balance profitability and competitiveness.

## Recommendations :
1. Regularly monitor competitor prices.
2. Maintain a healthy profit margin without overpricing products.
3. Use promotional discounts during low-demand periods.
4. Adjust prices according to customer demand and market conditions.
5. Review pricing strategies periodically to maximize revenue.
